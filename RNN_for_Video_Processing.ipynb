```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
num_samples = 100
timesteps = 10
height, width, channels = 64, 64, 1
x_data = np.random.rand(num_samples, timesteps, height, width, channels)
y_data = np.random.randint(0, 3, size=(num_samples,)) # 3 activity classes
x_data = x_data.reshape(num_samples, timesteps, -1)
x_train, x_test = x_data[:80], x_data[80:]
y_train, y_test = y_data[:80], y_data[80:]
model = models.Sequential([
layers.LSTM(64, input_shape=(timesteps, height * width * channels)),
layers.Dense(32, activation='relu'),
layers.Dense(3, activation='softmax') # 3 activity classes
])
model.compile(optimizer='adam',
loss='sparse_categorical_crossentropy',
metrics=['accuracy'])
model.fit(x_train, y_train, epochs=5, batch_size=8, validation_split=0.2)
loss, acc = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {acc * 100:.2f}%")
sample = x_test[0:1]
predicted = model.predict(sample)
print("Predicted Class:", np.argmax(predicted))
```
